```python
import nflreadpy as nfl
from nflreadpy.config import update_config
import numpy as np
import pandas as pd

update_config(cache_mode="filesystem")
```

```python
# Retreive data - run this once
all_variables = [
    "game_id", "play_id", "season", "week", "season_type", "drive",
    "posteam", "defteam", "home_team", "away_team",
    "down", "qtr", "play_type", "qb_kneel", "qb_spike", "pass", "rush",
    "yardline_100", "ydstogo",
    "yards_gained", "touchdown", "td_team", "interception", "fumble_lost",
    "epa", "posteam_score_post", "defteam_score_post",
]

plays = nfl.load_pbp(seasons=range(2021, 2026)).select(all_variables).to_pandas()
print(plays.shape)
plays.to_csv("data_raw.csv", index=False)
```

```python
# Save core variables
pd.read_csv("data_raw.csv")
core_variables = ["game_id", "play_id", "season", "season_type", "drive", "posteam", "defteam",
             "down", "qtr", "play_type", "qb_kneel", "qb_spike", "pass", "rush",
             "yardline_100", "ydstogo", "yards_gained", "touchdown", "td_team", "interception", "fumble_lost"]
plays[core_variables].to_csv("data_core_raw.csv", index=False)
```
